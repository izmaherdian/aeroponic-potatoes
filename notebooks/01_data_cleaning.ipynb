```python
import os
import shutil
from collections import defaultdict

folder_asal = "D:\Pemrograman\camera_1\camera_1"
folder_tujuan = "D:\Pemrograman\DataBersih\camera_1"

os.makedirs(folder_tujuan, exist_ok=True)

# Dictionary: {tanggal: {jam: file}}
file_terpilih = defaultdict(dict)

# Baca semua file
for file in sorted(os.listdir(folder_asal)):  # pakai sorted biar urut
    if file.endswith(".jpg"):
        try:
            tanggal, waktu = file.split("_")
            jam, menit, detik = waktu.split(".")[0].split("-")

            # kalau jam ini belum ada di tanggal tsb, simpan file pertama
            if jam not in file_terpilih[tanggal]:
                file_terpilih[tanggal][jam] = file
        except ValueError:
            continue

# Salin hasil terpilih
jumlah = 0
for tanggal, jam_dict in file_terpilih.items():
    for jam, f in jam_dict.items():
        asal = os.path.join(folder_asal, f)
        tujuan = os.path.join(folder_tujuan, f)
        shutil.copy2(asal, tujuan)
        jumlah += 1

print("Selesai! Jumlah file terpilih:", jumlah)
```
